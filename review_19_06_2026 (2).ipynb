## Q1. even number filter

#### a data entry clerk needs to seperate even-numbered invoice IDs for batch recociliation.print all even invoice IDs from the list.

```python
invoice_IDs=[1023,1044,1057,1088,1091,1102,1115,1130]
```

```python
print("even invoice IDs")
for i in invoice_IDs:
    if i%2==0:
        print(i)
```

## Q2. movie rating bucket counter

#### a streaming platform wants to bucket movie ratings into poor (<5),average(5-7),good(7-9),excellent(>=9).count movie per bucket.

```python
ratings=[4.2,6.8,8.1,9.3,5.5,7.9,3.8,8.8,6.1,9.5]
```

```python
poor=0
average=0
good=0
excellent=0
for i in ratings:
    if i<5:
         poor +=1
    elif i<7:
          average +=1
    elif i<9:
          good+=1
    else:
          excellent+=1
print("poor:",poor)
print("average:",average)
print("good:",good)
print("excellent:",excellent)

```
